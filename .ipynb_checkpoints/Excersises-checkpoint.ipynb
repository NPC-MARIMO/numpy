```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],  # Student 1
    [19, 92, 88],  # Student 2
    [17, 76, 95],  # Student 3
    [18, 65, 70],  # Student 4
    [20, 90, 85]   # Student 5
])
```

```python
# get the shape of matrix
data.shape
```

```python
# find the avg age of students
ages = data[:,0]
sum_of_ages = np.sum(ages, 0)
total_students = len(ages)
average_age = sum_of_ages / total_students
average_age
```
