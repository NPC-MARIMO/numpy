```python
# indexing and slicing of vectors
```

```python
import numpy as np 
```

```python
arr = np.arange(11,21)
arr
```

```python
arr[0]
```

```python
arr[0:8]
```

```python
arr[2:]
```

```python
arr[1::2]
```

```python
# indexing and slicing of matrix
```

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[:,0]
```

```python
arr[1:3,2:]
```

```python
arr[0::2,0::3]
```

```python
arr = np.arange(11, 21)
arr
```

```python
bool_index = arr%2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
