```python
#Deep and Shallow Copy
```

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice * 10
slice # called shallow copy where you change the sliced part of array and it does not affect the real array
```

```python
a
```

```python
a[4:6] * 10
```

```python
a
```

```python
b = a.copy()
```

```python
b[0] = 99
b
```

```python
a
```

```python
b = a
```

```python
b[0] = 99
b # deep copy where changing the copy of array also changes the real array 
```

```python
a
```
